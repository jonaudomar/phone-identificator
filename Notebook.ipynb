Etape suivante : normaliser le numéro appelant:
- supprimer espace et tiret et lettres
- supprier le 0033 ou +33 si existant
- vérifier que le numéro comporte 10 chiffres

créer une fonction 'appelant_normalized' qui renvoit un numéro normalisé de 10 chiffres sans autre caractère


```python
def appelant_normalized(number: str) -> str:
    value = ''.join(c for c in number if c.isdigit())

    if value.startswith('33'):
        value = '0' + value[2:]

    if value.startswith('0033'):
        value = '0' + value[4:]

    return value
```

```python
def is_spam(appelant:str):
    """
    Ceci est une super fonction
    """
    spam = False

    for prefixe in prospection:
        if appelant.startswith(prefixe):
            spam = True
            break
    return 'spam' if spam else 'tout va bien'
```
